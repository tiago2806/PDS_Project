```python
data.drop_duplicates(inplace = True)
```

```python
split_index = data.index.to_series().apply(
    lambda s: re.split(r'[,;.\-_/]+', s.strip())  # split on any separator
)

# --- 2️⃣ Rejoin all but the last part as "city" ---
split_index = pd.DataFrame(split_index)
split_index.columns = ['parts']
split_index['parts'] = split_index['parts'].apply(lambda lst: [x.strip() for x in lst if x.strip()])

split_index['Country'] = split_index['parts'].apply(lambda lst: lst[-1] if len(lst) > 1 else None)
split_index['City'] = split_index['parts'].apply(lambda lst: ' '.join(lst[:-1]) if len(lst) > 1 else lst[0])

# --- 3️⃣ Fix reversed entries (where the first token looks like a country) ---
# Build a probable country set (based on repetition)
country_counts = split_index['Country'].value_counts()
probable_countries = set(country_counts[country_counts > 1].index)

split_index['Country'], split_index['City'] = zip(*split_index.apply(
    lambda row: (
        row['Country'], row['City']
    ) if row['Country'] in probable_countries or row['Country'] is None else (
        row['City'], row['Country']
    ), axis=1
))

# --- 4️⃣ Create MultiIndex ---
data.index = pd.MultiIndex.from_frame(split_index[['Country', 'City']], names=['Country', 'City'])
data.sort_index(inplace=True)

display(data)
```
